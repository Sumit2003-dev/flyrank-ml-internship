**Decision my work improves:** Which pages the content team should review first for
refresh, given limited reviewer time.

**Who acts on it:** A content reviewer/strategist, who gets a ranked list instead of
manually scanning thousands of pages.

**The action:** Reviewer opens the top of the ranked queue and checks each page's
reason code, then decides whether to refresh, expand, protect, prune, or monitor it.

**Cost of a wrong call:**
- False positive (flagged, but page is actually fine): wastes reviewer time — low but
  compounding cost.
- False negative (real decline missed): the page keeps losing visibility unnoticed,
  and search rankings can be slow to recover — higher cost.
  

```python
print("Decision: prioritize pages for review")
print("Actor: content reviewer")
print("Cost of false negative > cost of false positive (rankings slow to recover)")
```
